```python
from datetime import datetime
start_time = datetime(2022,6,14,11,0,0)
end_time  = datetime.now()                         
duration = end_time - start_time                        
duration_in_s = duration.total_seconds()
days    = divmod(duration_in_s, 86400)
hours   = divmod(days[1], 3600)
minutes = divmod(hours[1], 60)
seconds = divmod(minutes[1], 1)
print("Time between dates: %d days, %d hours, %d minutes and %d seconds" % (days[0], hours[0], minutes[0], seconds[0]))
```

```python
def calculate_time_interval(start_time, end_time):
    from datetime import datetime                     
    duration = end_time - start_time                        
    duration_in_s = duration.total_seconds()
    days    = divmod(duration_in_s, 86400)
    hours   = divmod(days[1], 3600)
    minutes = divmod(hours[1], 60)
    seconds = divmod(minutes[1], 1)
    time_interval = "Time Taken: %d days, %d hours, %d minutes, %d seconds" % (days[0], hours[0], minutes[0], seconds[0])
    return time_interval
```

```python
import requests
from scrapy import Selector
import us
result = requests.get("https://www.officialusa.com/names/Alexander-Aab/")
response = Selector(result)
```

```python
persons = response.css(".detail-block__main-items .person")
```

```python
def possible_surname(response):
    data = response.css(".detail-block__main-item-summary-list li")   
    surnames = []
    for field in data:
        name = field.xpath("normalize-space()").get()
        surnames.append(name)
    return surnames
```

```python
for i in persons:
    name = i.xpath(".//span[@itemprop='name']/text()").get()
    telephones = i.xpath(".//span[@itemprop='telephone']/text()").getall()
    person_age = i.xpath("./@data-age").get()
    city = i.xpath(".//span[@itemprop='addressLocality']/text()").get()
    Region = i.xpath(".//span[@itemprop='addressRegion']/text()").get()
    State = us.states.lookup(Region)
    su_nmaes = possible_surname(response)
    print(name)
    print(telephones)
    print(person_age)
    print(city)
    print(Region)
    print(State)
    print(su_nmaes)
    print("*******************")
```
